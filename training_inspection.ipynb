```python
import os
# Set TF log level before importing TensorFlow
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3' 
import numpy as np
import pandas as pd
import librosa
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Conv2D, MaxPooling2D, GlobalAveragePooling2D, Dense,
    MultiHeadAttention, LayerNormalization, Dropout, Add, Flatten,
    TimeDistributed, GlobalAveragePooling1D, Reshape, Conv1D,
    Concatenate
)
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.utils import to_categorical
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import warnings

# Suppress warnings
warnings.filterwarnings('ignore')
tf.get_logger().setLevel('ERROR')

# --- 1. Configuration ---

# Dataset paths (as specified by user)
BASE_PATH = "/kaggle/input/mrs-affia"
RGB_PATH = os.path.join(BASE_PATH, "RGB", "RGB")
SI_PATH = os.path.join(BASE_PATH, "SI", "SI")
MEL_PATH = os.path.join(BASE_PATH, "mel", "mel") # This contains .wav files
CLASSES = ["none", "weak", "medium", "strong"]

# Model & Preprocessing Parameters
IMG_SIZE = (224, 224)
N_CLASSES = len(CLASSES)
BATCH_SIZE = 32
EPOCHS = 100 # Increased epochs, with early stopping
PATIENCE = 15

# Audio parameters (inspired by Iqbal et al. and MFFFI)
SR = 22050
DURATION = 3.0
N_MELS = 64
N_FFT = 2048
HOP_LENGTH = 512
MAX_TIME_STEPS = 130 # Pre-calculated: (SR * DURATION) / HOP_LENGTH + 1

# --- 2. Data Loading and Preparation ---

def find_and_match_files():
    """
    Finds all files, matches them by name across the three modalities,
    and creates a DataFrame.
    """
    data = []
    
    # Create dictionaries mapping filenames (without extension) to full paths
    def get_file_map(base_dir):
        file_map = {}
        for label in CLASSES:
            class_dir = os.path.join(base_dir, label)
            if not os.path.isdir(class_dir):
                print(f"Warning: Directory not found {class_dir}")
                continue
            for f in os.listdir(class_dir):
                filename_only = os.path.splitext(f)[0] # e.g., "RGB_0" or "mel_0"
                
                # --- FIX: Create a generic key like "strong_0" ---
                file_id_parts = filename_only.split('_')
                if len(file_id_parts) > 0:
                    file_id = file_id_parts[-1] # Get the number, e.g., "0"
                    generic_key = f"{label}_{file_id}"
                    file_map[generic_key] = os.path.join(class_dir, f)
                # --------------------------------------------------
        return file_map

    print("Mapping RGB files...")
    rgb_map = get_file_map(RGB_PATH)
    print("Mapping SI files...")
    si_map = get_file_map(SI_PATH)
    print("Mapping Audio (.wav) files...")
    mel_map = get_file_map(MEL_PATH)

    print(f"Found: {len(rgb_map)} RGB, {len(si_map)} SI, {len(mel_map)} Audio files.")

    # Match files
    for label in CLASSES:
        # We iterate by number, assuming filenames like RGB_0.jpg, SI_0.jpg, mel_0.wav
        # Let's find the max index for this label based on one modality (e.g., RGB)
        # --- FIX: Use the generic keys from rgb_map ---
        label_files = [k for k in rgb_map.keys() if k.startswith(label)]
        
        for key in label_files: # key is now, e.g., "strong_0"
            if key in si_map and key in mel_map: # This check will now succeed
                data.append({
                    "rgb_path": rgb_map[key],
                    "si_path": si_map[key],
                    "mel_path": mel_map[key],
                    "label": label
                })
            else:
                print(f"Warning: Missing match for key {key}")
        # ------------------------------------------------

    df = pd.DataFrame(data)
    print(f"Successfully matched {len(df)} tri-modal samples.")
    return df

# Create DataFrame
data_df = find_and_match_files()

if data_df.empty:
    print("Error: No data loaded. Check Kaggle paths and file matching logic.")
else:
    # Encode labels
    label_encoder = LabelEncoder()
    data_df["label_encoded"] = label_encoder.fit_transform(data_df["label"])
    
    # Split data
    train_df, val_df = train_test_split(
        data_df,
        test_size=0.2,
        random_state=42,
        stratify=data_df["label_encoded"]
    )
    
    print(f"Training samples: {len(train_df)}, Validation samples: {len(val_df)}")

# --- 3. Preprocessing Functions (for tf.data) ---

def load_rgb_image(path):
    img = tf.io.read_file(path)
    img = tf.image.decode_jpeg(img, channels=3)
    img = tf.image.resize(img, IMG_SIZE)
    img = img / 255.0  # Normalize to [0, 1]
    return img

def load_si_image(path):
    img = tf.io.read_file(path)
    # Assuming sonar images are JPG, adjust if PNG
    img = tf.image.decode_jpeg(img, channels=1) 
    img = tf.image.resize(img, IMG_SIZE)
    img = img / 255.0  # Normalize to [0, 1]
    return img

def load_audio_spectrogram(path):
    """
    Loads a .wav file, pads/truncates, and computes a Mel spectrogram.
    This function will be wrapped by tf.py_function.
    """
    try:
        # --- FIX: Decode EagerTensor to string ---
        # Decode path from tensor
        path = path.numpy().decode('utf-8')
        # -----------------------------------------
        
        y, _ = librosa.load(path, sr=SR, duration=DURATION)
        
        # Pad/Truncate
        target_len = int(SR * DURATION)
        if len(y) < target_len:
            y = np.pad(y, (0, target_len - len(y)), mode='constant')
        else:
            y = y[:target_len]
        
        # Compute Mel spectrogram
        mel = librosa.feature.melspectrogram(
            y=y, sr=SR, n_mels=N_MELS, n_fft=N_FFT, hop_length=HOP_LENGTH
        )
        mel_db = librosa.power_to_db(mel, ref=np.max)
        
        # Pad time steps to be uniform (MAX_TIME_STEPS)
        if mel_db.shape[1] < MAX_TIME_STEPS:
            pad_width = MAX_TIME_STEPS - mel_db.shape[1]
            mel_db = np.pad(mel_db, ((0, 0), (0, pad_width)), mode='constant')
        else:
            mel_db = mel_db[:, :MAX_TIME_STEPS]

        # Normalize
        mel_db = (mel_db - np.mean(mel_db)) / (np.std(mel_db) + 1e-6)
        
        # Transpose to (Time, Freq) for Transformer
        mel_db = mel_db.T 
        
        return mel_db.astype(np.float32)
        
    except Exception as e:
        print(f"Error loading audio file {path}: {e}")
        return np.zeros((MAX_TIME_STEPS, N_MELS), dtype=np.float32)

@tf.function
def tf_load_audio(path):
    """Wraps the numpy/librosa function for use in tf.data"""
    [mel_db,] = tf.py_function(load_audio_spectrogram, [path], [tf.float32])
    mel_db.set_shape((MAX_TIME_STEPS, N_MELS))
    return mel_db

def load_and_preprocess(rgb_path, si_path, mel_path, label):
    """
    The main mapping function for tf.data.Dataset
    """
    rgb_img = load_rgb_image(rgb_path)
    si_img = load_si_image(si_path)
    audio_mel = tf_load_audio(mel_path)
    
    label_one_hot = tf.one_hot(label, depth=N_CLASSES)
    
    return (rgb_img, si_img, audio_mel), label_one_hot

# --- 4. tf.data.Dataset Pipeline ---

def create_dataset(df):
    dataset = tf.data.Dataset.from_tensor_slices((
        df["rgb_path"].values,
        df["si_path"].values,
        df["mel_path"].values,
        df["label_encoded"].values.astype(np.int32)
    ))
    
    dataset = dataset.shuffle(buffer_size=len(df))
    dataset = dataset.map(
        load_and_preprocess, 
        num_parallel_calls=tf.data.AUTOTUNE
    )
    dataset = dataset.batch(BATCH_SIZE)
    dataset = dataset.prefetch(buffer_size=tf.data.AUTOTUNE)
    return dataset

if not data_df.empty:
    train_dataset = create_dataset(train_df)
    val_dataset = create_dataset(val_df)
    
    # Let's check the output shape
    print("\n--- Dataset Output Shapes ---")
    for (rgb, si, audio), label in train_dataset.take(1):
        print(f"RGB Shape:    {rgb.shape}")
        print(f"SI Shape:     {si.shape}")
        print(f"Audio Shape:  {audio.shape}")
        print(f"Label Shape:  {label.shape}")

# --- 5. Model Architecture (UAT-FFIA) ---

def build_uat_ffia_model(
    img_shape=(224, 224, 3), 
    si_shape=(224, 224, 1), 
    audio_shape=(130, 64), 
    num_classes=4,
    d_model=64 # Dimension for feature projection and attention
):
    """
    Builds the proposed Efficient Tri-Modal Fusion (UAT-FFIA) model.
    """
    
    # === Branch 1: Lightweight RGB CNN ===
    rgb_input = Input(shape=img_shape, name='rgb_input')
    x_rgb = Conv2D(16, (3, 3), activation='relu', padding='same')(rgb_input)
    x_rgb = MaxPooling2D((2, 2))(x_rgb)
    x_rgb = Conv2D(32, (3, 3), activation='relu', padding='same')(x_rgb)
    x_rgb = MaxPooling2D((2, 2))(x_rgb)
    rgb_features_raw = GlobalAveragePooling2D(name='rgb_gap')(x_rgb)
    # Project features to d_model
    rgb_features = Dense(d_model, activation='relu', name='rgb_projection')(rgb_features_raw)

    # === Branch 2: Lightweight Sonar (SI) CNN ===
    si_input = Input(shape=si_shape, name='si_input')
    x_si = Conv2D(16, (3, 3), activation='relu', padding='same')(si_input)
    x_si = MaxPooling2D((2, 2))(x_si)
    x_si = Conv2D(32, (3, 3), activation='relu', padding='same')(x_si)
    x_si = MaxPooling2D((2, 2))(x_si)
    si_features_raw = GlobalAveragePooling2D(name='si_gap')(x_si)
    # Project features to d_model
    si_features = Dense(d_model, activation='relu', name='si_projection')(si_features_raw)

    # === Branch 3: Small Audio Transformer ===
    audio_input = Input(shape=audio_shape, name='audio_input')
    # Normalize input sequence
    x_audio = LayerNormalization(name='audio_norm_in')(audio_input)
    
    # --- Transformer Block 1 ---
    # Self-Attention
    att_audio = MultiHeadAttention(
        num_heads=4, key_dim=d_model, name='audio_mha_1'
    )(x_audio, x_audio)
    x_audio = Add(name='audio_add_1')([x_audio, att_audio])
    x_audio = LayerNormalization(name='audio_norm_1')(x_audio)
    
    # Feed-Forward Network
    ffn_audio = Dense(d_model * 2, activation='relu', name='audio_ffn_1_dense_1')(x_audio)
    ffn_audio = Dense(d_model, name='audio_ffn_1_dense_2')(ffn_audio)
    x_audio = Add(name='audio_add_2')([x_audio, ffn_audio])
    x_audio = LayerNormalization(name='audio_norm_2')(x_audio)
    
    # Pool features over time
    audio_features = GlobalAveragePooling1D(name='audio_gap')(x_audio)
    # audio_features is already (d_model), no projection needed

    # === Fusion: Cross-Modal Attention ===
    
    # --- FIX: Replace tf.stack with Keras layers ---
    # 1. Reshape features to (BatchSize, 1, d_model) to create a sequence
    rgb_seq = Reshape((1, d_model), name='rgb_seq')(rgb_features)
    si_seq = Reshape((1, d_model), name='si_seq')(si_features)
    audio_seq = Reshape((1, d_model), name='audio_seq')(audio_features)

    # 2. Concatenate into a sequence of length 3
    fused_sequence = Concatenate(axis=1, name='modal_stack')([rgb_seq, si_seq, audio_seq])
    # Shape is now (BatchSize, 3, d_model)
    # -----------------------------------------------
    
    # 3. Apply cross-modal attention (self-attention on the modal sequence)
    # This allows modalities to "attend" to each other
    x_fused = LayerNormalization(name='fusion_norm_in')(fused_sequence)
    att_fused = MultiHeadAttention(
        num_heads=4, key_dim=d_model, name='fusion_mha'
    )(x_fused, x_fused)
    x_fused = Add(name='fusion_add_1')([x_fused, att_fused])
    x_fused = LayerNormalization(name='fusion_norm_1')(x_fused)

    # 4. Feed-Forward Network for fused features
    ffn_fused = Dense(d_model * 2, activation='relu', name='fusion_ffn_dense_1')(x_fused)
    ffn_fused = Dense(d_model, name='fusion_ffn_dense_2')(ffn_fused)
    x_fused = Add(name='fusion_add_2')([x_fused, ffn_fused])
    x_fused = LayerNormalization(name='fusion_norm_2')(x_fused)
    
    # 5. Flatten the attended modal features
    fused_features = Flatten(name='fusion_flatten')(x_fused)

    # === Classifier Head ===
    x = Dense(d_model, activation='relu', name='classifier_dense_1')(fused_features)
    x = Dropout(0.3, name='classifier_dropout')(x)
    output = Dense(num_classes, activation='softmax', name='output')(x)

    # --- Build Model ---
    model = Model(
        inputs=[rgb_input, si_input, audio_input],
        outputs=output,
        name="UAT_FFIA_Model"
    )
    
    return model

# Build and compile the model
model = build_uat_ffia_model(
    img_shape=IMG_SIZE + (3,),
    si_shape=IMG_SIZE + (1,),
    audio_shape=(MAX_TIME_STEPS, N_MELS),
    num_classes=N_CLASSES
)

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()

# Check parameter count against target (< 1.5M)
total_params = model.count_params()
print(f"\nTotal Parameters: {total_params / 1_000_000:.2f} M")
if total_params < 1_500_000:
    print("Parameter count is UNDER the 1.5M target. Success!")
else:
    print("Warning: Parameter count is OVER the 1.5M target.")


# --- 6. Training ---

if not data_df.empty:
    print("\n--- Starting Model Training ---")
    
    callbacks = [
        EarlyStopping(
            monitor='val_loss', 
            patience=PATIENCE, 
            restore_best_weights=True,
            verbose=1
        ),
        ModelCheckpoint(
            filepath='uat_ffia_best.keras',
            monitor='val_accuracy',
            save_best_only=True,
            verbose=1
        ),
        ReduceLROnPlateau(
            monitor='val_loss',
            factor=0.2,
            patience=5,
            min_lr=1e-6,
            verbose=1
        )
    ]
    
    history = model.fit(
        train_dataset,
        validation_data=val_dataset,
        epochs=EPOCHS,
        callbacks=callbacks
    )
    
    print("\n--- Training Finished ---")

    # --- 7. Evaluation ---
    print("\n--- Evaluating Model ---")
    
    # Load best model
    model.load_weights('uat_ffia_best.keras')
    
    # Evaluate on validation set
    loss, accuracy = model.evaluate(val_dataset)
    print(f"Validation Loss:     {loss:.4f}")
    print(f"Validation Accuracy: {accuracy * 100:.2f}%")
    
    # --- Classification Report and Confusion Matrix ---
    print("\n--- Classification Report ---")
    y_pred = []
    y_true = []
    
    for (rgb, si, audio), label in val_dataset:
        preds = model.predict((rgb, si, audio), verbose=0)
        y_pred.extend(np.argmax(preds, axis=1))
        y_true.extend(np.argmax(label, axis=1))
        
    print(classification_report(y_true, y_pred, target_names=CLASSES))
    
    print("--- Confusion Matrix ---")
    cm = confusion_matrix(y_true, y_pred)
    print(pd.DataFrame(cm, index=CLASSES, columns=CLASSES))
    
    # --- Plot Training History ---
    print("\n--- Plotting Training History ---")
    
    try:
        plt.figure(figsize=(12, 4))
        
        # Plot Accuracy
        plt.subplot(1, 2, 1)
        plt.plot(history.history['accuracy'], label='Training Accuracy')
        plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
        plt.title('Model Accuracy')
        plt.xlabel('Epoch')
        plt.ylabel('Accuracy')
        plt.legend()
        
        # Plot Loss
        plt.subplot(1, 2, 2)
        plt.plot(history.history['loss'], label='Training Loss')
        plt.plot(history.history['val_loss'], label='Validation Loss')
        plt.title('Model Loss')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.legend()
        
        plt.tight_layout()
        plt.savefig('training_history.png')
        print("Training history plot saved as 'training_history.png'")
        
    except Exception as e:
        print(f"Could not plot history: {e}")

else:
    print("\nSkipping training and evaluation as no data was loaded.")
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import librosa
import librosa.display
from PIL import Image

# --- CONFIGURATION ---
# Update this path to point to your dataset location
BASE_PATH = "/kaggle/input/mrs-affia" 
CLASSES = ["strong", "medium", "weak", "none"]
OUTPUT_DIR = "./paper_figures"

# Create output directory if it doesn't exist
os.makedirs(OUTPUT_DIR, exist_ok=True)
def generate_paper_figure_1(base_path, classes):
    print("Generating Composite Figure 1...")
    
    # Create a grid: 3 Rows (RGB, SI, Audio) x 4 Columns (Classes)
    fig, axes = plt.subplots(3, 4, figsize=(16, 10))
    
    # Row Labels
    row_names = ["Visual (RGB)", "Sonar (SI)", "Audio (Mel)"]
    
    for col, label in enumerate(classes):
        print(f"Processing class: {label}...")
        
        # --- 1. Load RGB Image ---
        rgb_dir = os.path.join(base_path, "RGB", "RGB", label)
        # Get the first file found
        rgb_file = os.listdir(rgb_dir)[0] 
        rgb_img = Image.open(os.path.join(rgb_dir, rgb_file))
        
        axes[0, col].imshow(rgb_img)
        axes[0, col].axis('off')
        axes[0, col].set_title(label.capitalize(), fontsize=14, fontweight='bold')

        # --- 2. Load Sonar (SI) Image ---
        si_dir = os.path.join(base_path, "SI", "SI", label)
        si_file = os.listdir(si_dir)[0]
        si_img = Image.open(os.path.join(si_dir, si_file)).convert('L') # Force grayscale
        
        axes[1, col].imshow(si_img, cmap='gray')
        axes[1, col].axis('off')

        # --- 3. Generate Mel Spectrogram ---
        # We process this exactly as your Model Architecture description states
        mel_dir = os.path.join(base_path, "mel", "mel", label)
        wav_file = os.listdir(mel_dir)[0]
        
        # Load audio
        y, sr = librosa.load(os.path.join(mel_dir, wav_file), sr=22050, duration=3.0)
        
        # Compute Mel Spectrogram (Matching your Methodology parameters)
        S = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=64, n_fft=2048, hop_length=512)
        S_dB = librosa.power_to_db(S, ref=np.max)
        
        # Display
        img = librosa.display.specshow(S_dB, sr=sr, hop_length=512, x_axis='time', y_axis='mel', 
                                     ax=axes[2, col], cmap='inferno')
        axes[2, col].axis('off') # Turn off axis for clean look

    # Add Row Labels to the left
    for i, row_name in enumerate(row_names):
        axes[i, 0].annotate(row_name, xy=(-0.1, 0.5), xycoords='axes fraction', 
                            fontsize=14, fontweight='bold', ha='right', va='center', rotation=90)

    plt.tight_layout()
    save_path = os.path.join(OUTPUT_DIR, "Figure_1_Composite.png")
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    print(f"✅ Figure 1 saved to: {save_path}")
    plt.show()

# Run Option A
generate_paper_figure_1(BASE_PATH, CLASSES)
```

```python
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

def generate_figure_2(audio_path=None):
    # --- 1. Configuration ---
    SR = 22050
    DURATION = 3.0
    N_FFT = 2048
    HOP_LENGTH = 512
    N_MELS = 64

    # --- 2. Load Audio ---
    if audio_path is None:
        # Use a demo file if no path provided
        audio_path = librosa.ex('trumpet')
    
    y, sr = librosa.load(audio_path, sr=SR, duration=DURATION)

    # --- 3. Generate Mel Spectrogram ---
    S = librosa.feature.melspectrogram(
        y=y, 
        sr=sr, 
        n_fft=N_FFT, 
        hop_length=HOP_LENGTH, 
        n_mels=N_MELS
    )
    S_dB = librosa.power_to_db(S, ref=np.max)

    # --- 4. Plotting ---
    fig, ax = plt.subplots(nrows=2, ncols=1, figsize=(10, 8), sharex=True)

    # Plot 1: Waveform
    librosa.display.waveshow(y, sr=sr, ax=ax[0], color='blue', alpha=0.6)
    ax[0].set_title("1. Original Audio Waveform", fontsize=14, fontweight='bold')
    ax[0].set_ylabel("Amplitude")
    ax[0].label_outer()

    # Plot 2: Mel Spectrogram (No Colorbar)
    img = librosa.display.specshow(
        S_dB, 
        sr=sr, 
        hop_length=HOP_LENGTH, 
        x_axis='time', 
        y_axis='mel', 
        fmax=8000, 
        ax=ax[1], 
        cmap='inferno'
    )
    
    ax[1].set_title("2. Mel Spectrogram (dB Scale)", fontsize=14, fontweight='bold')
    ax[1].set_xlabel("Time (s)")
    ax[1].set_ylabel("Mel Frequency")

    plt.tight_layout()
    
    # --- 5. Save the Figure ---
    save_name = 'figure2_audio_preprocessing.png'
    plt.savefig(save_name, dpi=300, bbox_inches='tight')
    print(f"✅ Image saved successfully as: {save_name}")
    
    plt.show()

# --- Run ---
# You can pass your specific file path here if you want
# generate_figure_2("/kaggle/input/mrs-affia/mel/mel/strong/strong_10.wav")
generate_figure_2()
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import librosa
import librosa.display
from PIL import Image
import random

# --- CONFIGURATION ---
BASE_PATH = "/kaggle/input/mrs-affia" 
CLASSES = ["strong", "medium", "weak", "none"]
MODALITIES = ["RGB (Visual)", "SI (Sonar)", "Audio (Mel)"]
OUTPUT_DIR = "./paper_figures"
IMG_SIZE = (250, 200)

os.makedirs(OUTPUT_DIR, exist_ok=True)

def get_random_synced_files(base_path, label):
    """
    Finds a random sample ID that exists in RGB, SI, and Mel folders
    to ensure the visualization represents the SAME event.
    """
    rgb_dir = os.path.join(base_path, "RGB", "RGB", label)
    si_dir = os.path.join(base_path, "SI", "SI", label)
    mel_dir = os.path.join(base_path, "mel", "mel", label)
    
    # Helper to map IDs to filenames
    # Assumes format like "class_ID.ext", taking the last part after '_' as ID
    def get_id_map(directory):
        files = os.listdir(directory)
        id_map = {}
        for f in files:
            # Extract ID: "strong_10.jpg" -> "10"
            file_id = os.path.splitext(f)[0].split('_')[-1]
            id_map[file_id] = f
        return id_map

    rgb_map = get_id_map(rgb_dir)
    si_map = get_id_map(si_dir)
    mel_map = get_id_map(mel_dir)
    
    # Find intersection (IDs present in ALL 3 modalities)
    common_ids = list(set(rgb_map.keys()) & set(si_map.keys()) & set(mel_map.keys()))
    
    if not common_ids:
        print(f"Warning: No matching IDs found for class {label}. Picking random independent files.")
        return os.listdir(rgb_dir)[0], os.listdir(si_dir)[0], os.listdir(mel_dir)[0]
    
    # Pick one random ID
    selected_id = random.choice(common_ids)
    print(f"  Selected Sample ID for {label}: {selected_id}")
    
    return rgb_map[selected_id], si_map[selected_id], mel_map[selected_id]

def generate_synced_grid_figure(base_path, classes):
    print("Generating Synchronized Data Grid...")
    
    fig, axes = plt.subplots(4, 3, figsize=(12, 16))
    plt.subplots_adjust(wspace=0.05, hspace=0.05)

    # Set Column Headers
    for i, mod in enumerate(MODALITIES):
        axes[0, i].set_title(mod, fontsize=18, fontweight='bold', pad=15)

    for row, label in enumerate(classes):
        print(f"Processing class: {label}...")
        
        # Get synchronized filenames
        f_rgb, f_si, f_mel = get_random_synced_files(base_path, label)
        
        # --- 1. RGB (Visual) ---
        rgb_path = os.path.join(base_path, "RGB", "RGB", label, f_rgb)
        rgb_img = Image.open(rgb_path).resize(IMG_SIZE)
        
        axes[row, 0].imshow(rgb_img)
        axes[row, 0].axis('off')

        # --- 2. SI (Sonar) - COLORED ---
        si_path = os.path.join(base_path, "SI", "SI", label, f_si)
        si_img = Image.open(si_path).convert('L').resize(IMG_SIZE)
        
        axes[row, 1].imshow(si_img, cmap='viridis') 
        axes[row, 1].axis('off')

        # --- 3. Audio (Mel Spectrogram) ---
        mel_path = os.path.join(base_path, "mel", "mel", label, f_mel)
        
        # Load and process
        y, sr = librosa.load(mel_path, sr=22050, duration=3.0)
        S = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=64, n_fft=2048, hop_length=512)
        S_dB = librosa.power_to_db(S, ref=np.max)
        
        # Plot
        librosa.display.specshow(S_dB, sr=sr, hop_length=512, x_axis='time', y_axis='mel', 
                                 ax=axes[row, 2], cmap='inferno')
        axes[row, 2].axis('off')

        # --- Add Row Labels ---
        axes[row, 0].text(-0.1, 0.5, label.capitalize(), transform=axes[row, 0].transAxes, 
                          fontsize=18, fontweight='bold', ha='right', va='center', rotation=90)

        # --- Add Grid Borders ---
        for col in range(3):
            rect = patches.Rectangle((0, 0), 1, 1, linewidth=2, edgecolor='black', facecolor='none', 
                                     transform=axes[row, col].transAxes)
            axes[row, col].add_patch(rect)

    save_path = os.path.join(OUTPUT_DIR, "Figure_1_Grid_Synced.png")
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    print(f"✅ Synchronized Figure saved to: {save_path}")
    plt.show()

generate_synced_grid_figure(BASE_PATH, CLASSES)
```
